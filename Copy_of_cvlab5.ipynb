```python
import cv2
import numpy as np
import matplotlib.pyplot as plt # Import for displaying images

rmin = 100 # Corrected assignment
rmax = 200 # Corrected assignment

# Load the image directly as grayscale
img = cv2.imread("/content/kl12.jpg", cv2.IMREAD_GRAYSCALE)

# Check if image was loaded
if img is None:
    print("Error: Could not load image. Please check the path.")
else:
    slicing_with_bg = img.copy() # Create a grayscale copy
    # slicing_wo_bg is a binary image (white for pixels in range, black otherwise)
    slicing_wo_bg = np.zeros(img.shape, dtype=np.uint8) # Initialized as grayscale

    # Loop through pixels
    for i in range(img.shape[0]):
      for j in range(img.shape[1]):
        pixel_value = img[i, j] # Get grayscale intensity for comparison from img

        if rmin <= pixel_value <= rmax:
          slicing_with_bg[i, j] = 255
          slicing_wo_bg[i, j] = 255
        else:
          slicing_wo_bg[i, j] = 0

    # Display images using matplotlib
    plt.figure(figsize=(15, 5))

    plt.subplot(1, 3, 1)
    plt.imshow(img, cmap='gray') # Display grayscale original image
    plt.title('Original Grayscale Image')
    plt.axis('off')

    plt.subplot(1, 3, 2)
    plt.imshow(slicing_with_bg, cmap='gray') # Display grayscale slicing_with_bg
    plt.title('Slicing with Background (Grayscale)')
    plt.axis('off')

    plt.subplot(1, 3, 3)
    plt.imshow(slicing_wo_bg, cmap='gray') # Display grayscale slicing_wo_bg
    plt.title('Slicing without Background (Grayscale)')
    plt.axis('off')

    plt.show()
```
