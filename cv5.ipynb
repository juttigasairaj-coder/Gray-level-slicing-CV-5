```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

#load the image in grayscale
img = cv2.imread("/content/wp9040584.jpg",cv2.IMREAD_GRAYSCALE)

#check if image loaded successfully
if img is None:
  print("Error loading image")
else:
  print("Image loaded successfully")

  #set up the plot figure
  plt.figure(figsize=(12,10))

  #Display original image
  plt.subplot(3,3,1)
  plt.imshow(img,cmap='gray')
  plt.title('Original Image')
  plt.axis('off')

  #loop through all 8 bit plane
  for i in range(8):
    #extract the i-th bit plane
    #shift bits right by 'i' and perform bitwise and with 1 bit_plane = (img >> i)&1
    vis_plane=bit_plane * 255

    #plotting
    plt.subplot(3,3,i+2)
    plt.imshow(vis_plane,cmap='gray')
    plt.title(f'Bit Plane {i}')
    plt.axis('off')
```
